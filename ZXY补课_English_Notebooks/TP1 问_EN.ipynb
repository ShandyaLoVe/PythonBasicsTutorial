The terminal and Python solve different layers of the same problem. Bash organises the project and its files; Python represents and processes the data inside those files. In both environments, reliable work depends on the same habit: inspect the current state before changing it, predict an outcome, run a small step, and verify the result.  
终端与 Python 处理的是同一任务的不同层面：Bash 负责项目和文件，Python 负责数据表达与计算。二者共同的可靠习惯是：修改前先检查状态，运行前先预测，小步执行，并立即验证结果。

### Self-check｜自我检查

- Can I explain the difference between `pwd`, `ls`, and `cd`?
- Can I distinguish an integer, float, string, complex number, and list?
- Can I identify the exception type from the final line of a traceback?
- Can I choose appropriate `range` arguments and explain its excluded endpoint?
- Can I use a loop and `enumerate` to produce a clearly labelled report?

如果以上任一问题还不能独立回答，请回到对应小节，先用一个更小的例子复现，再继续天气记录综合任务。
