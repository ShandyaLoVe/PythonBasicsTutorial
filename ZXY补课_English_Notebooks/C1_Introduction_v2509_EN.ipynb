### 2.1 - Lists｜2.1 - 列表


````python
colors = ["red", "green", "blue", 77, 99]

print("First item:", colors[0])
print("Last item:", colors[-1])
print("Slice from index 1:", colors[1:])

colors[1] = "cyan"       # replace an item
colors.append("black")   # add an item at the end

print()
removed_color = colors.pop(0)

# print("Updated list:", colors)
print("Removed item:", removed_color)
# print("Number of items:", len(colors))
````

### Complete list operations retained｜完整列表操作

Lists can mix different types. The following example retains replacement, `del`, `pop`, `insert`, `append`, indexing, and `len`. Remember that indices start at 0 and negative indices count from the end.  
列表可以混合不同类型。下面完整保留替换、`del`、`pop`、`insert`、`append`、索引和 `len`。请记住，索引从 0 开始，负索引从末尾向前计数。


````python
mixed_values = ["red", 18, "green", 22, "blue", 43.5, 3 < 5, True, 28.3]
print("Initial heterogeneous list:", mixed_values)

print("First item:", mixed_values[0])
print("Last item:", mixed_values[-1])


mixed_values[3] = "cyan"           # replace an item
del mixed_values[3]                # delete by index
removed_item = mixed_values.pop(3) # delete and return an item
mixed_values.insert(2, "black") # insert at a position
mixed_values.append(66.4)          # add at the end

print("Removed item:", removed_item)
print("Modified list:", mixed_values)
print("Length:", len(mixed_values))
````

### 2.3 - Dictionaries｜2.3 - 字典

A dictionary associates a unique **key** with a **value**. Unlike a list index, a key describes what the value means. Keys are often strings; values may have any type, including lists or other dictionaries.  
字典将独特的**键**与**值**联系起来. 与列表索引不同，一个密钥会描述值的含义. 键通常是字符串；值可能有任意类型，包括列表或其他字典。

Use a dictionary when data has labels such as `mass`, `unit`, or `operator`.  
当数据有`mass`,`unit`，或`operator`等标签时使用字典。


````python


experiment = {
    "name": "tensile test",
    "sample_id": 12,
    "force_N": [0.0, 12.5, 25.1],
    "validated": False,
}

print(experiment["name"])              # access a value
experiment["validated"] = True         # update a value
experiment["temperature_C"] = 21.4     # add a key-value pair
print(experiment)
````

Useful dictionary operations:  
有用的字典操作 :

- `data[key]` returns the value, but raises `KeyError` if the key is absent;  
  `data[key]` 返回该值，但如果缺少密钥，则提高 `KeyError` ;
- `data.get(key, default)` safely returns a default value for a missing key;  
  `data.get(key, default)` 安全返回缺失密钥的默认值；
- `data.keys()`, `data.values()`, and `data.items()` provide iterable views;  
  `data.keys()`、`data.values()`和`data.items()`提供了可使用的观点；
- `key in data` tests whether a key exists.  
  `key in data`测试是否存在密钥。


````python
print("Unknown location:", experiment.get("location", "not specified"))

for key, value in experiment.items():
    print(f"{key}: {value}")
````

In general, we use control statements in scripts. Instructions can be repeated (loops) or executed conditionally (tests).  
一般来说，我们在脚本中使用控制语句. 指令可以重复(loops)或有条件地执行(测试).

### 3.1 - Tests｜3.1 - 条件判断

The tests are done with the keywords `if`, `elif` and `else`.  
测试用关键词`if`、`elif`和`else`进行。

- Comparators are classic  `>`, `<`, `>=` (greater than or equal), `<=` (less than or equal), `==` (equality) and `!=` (different).  
  比较器是经典的`>`、`<`、`>=`(大于或等于)、`<=`(小于或等于)、`==`(平等)和`!=`(不同)。

以下是使用的例子：


````python
x = 7
if x < 0:
    print("x is strictly negative")
elif x % 2 != 0:  # % returns the remainder of a division
    print("x is positive and odd")
else:
    print("x is not negative and is even")    
````

### 3.2 - Loops｜3.2 - 循环

A `for` loop visits the items of an iterable. Prefer looping directly over values when you do not need an index.  
一个`for`环路访问一个可移动的物品。 在不需要索引时，更倾向于直接绕过值。


````python
colors = ["red", "green", "blue"]

for color in colors:
    print(color)

print("The loop is complete.")
````

A string is also iterable: a loop visits one character at a time. Strings, like tuples, are immutable.  
一个字符串也是可执行的：循环一次访问一个字符. 弦，如 tu，不可变。


````python
first_name = "Josephine"
for letter in first_name:
    print(letter)

print("Type:", type(first_name))
````

When integer indices are needed, use the iterable produced by `range`:  
当需要整数指数时，使用`range`生产的可更新指数：

- `range(stop)` generates integers from 0 to stop-1  
  `range(stop)` 生成整数从 0 到 stop-1
- `range(start,stop)` generates integers from start to stop-1  
  `range(start,stop)` 生成整数从开始到停止 - 页：1
- `range(start, stop, step)` generates integers from `start` toward, but excluding, `stop`.  
  `range(start, stop, step)`从`start`向`stop`生成整数，但不包括。

The `break` statement ends a loop early.  
`break`语句提前结束循环。


````python
for i in range(5):
    print(i)
print("Starts at 0 and ends at 4!")
````

````python
for i in range(2,17,3):
    if i>11:
        break  
    print(i, end=" ") # The end=" " replaces the newline in print() with a space.
    
print("\nThe loop is interrupted for i strictly greater than 11") #\n to go to the line
print("Value of i after exiting the loop: ", i)
````

Use `enumerate` when you need both the position and the value. Its `start` argument can make display numbering begin at 1.  
在需要位置和值时使用 `enumerate`。 它的`start`参数可以使显示编号从1开始。


````python
colors = ["red", "green", "blue"]
for number, color in enumerate(colors, start=1):
    print(f"Color no. {number} is {color}")
````

## 4 - Terminal input and text files｜4 - 终端输入和文本文件

When we write Python code for a user, we sometimes want this person to be able to configure the code by giving the value of certain parameters.  
当我们为用户写 Python 代码时，我们有时希望这个人能够通过给出某些参数的值来配置代码。

If we assume that this person uses the code in a terminal, we use the function `input` with the following syntax:  
如果我们假设此人在终端中使用代码，我们使用函数`input`，并使用以下语法：

```text
variable = input(prompt_text)
```
where  
地点

- `prompt_text` is the instruction displayed to the user;  
  `prompt_text`是向用户显示的指令；
- `variable` is the variable in which the result is stored.  
  `variable`是结果存储的变量。

> **Caution:** `input` returns only character strings (type `str`). If we want to use `variable` like a number, it must be converted first.  
> ** Caature:** `input`只返回字符串(类型为`str`). 如果我们想像数字一样使用`variable`，它必须先转换。

To write to the terminal, we use `print` as already seen above.  
要给终端写信，我们使用`print`，如上所述。

Below is an example.  
下面是一个例子。


````python
try:
    raw_value = input("Enter an integer: ")  # input always returns a string
except EOFError:
    raw_value = "4"  # non-interactive fallback for Run All / automated checks
    print("No interactive input available; using 4 for this demonstration.")
print("Before conversion:", raw_value, type(raw_value))

number = int(raw_value)
print("After conversion:", number, type(number))
print("Five times the number:", 5 * number)
````

Files allow data and results to persist after Python stops. Use `with open(...)` so that Python closes the file automatically, even if an error occurs. Common modes are `"r"` for reading and `"w"` for writing (and replacing existing content).  
文件允许Python停止后数据和结果持续. 使用 `with open(...)` 使 Python 自动关闭文件， 即使发生错误。 常见的模式是用于阅读的`"r"`和用于写作的`"w"`(和替换现有内容).


````python
message = "Measurement session complete.\nThree samples were tested."

with open("C1_session_note.txt", "w", encoding="utf-8") as output_file:
    output_file.write(message)

with open("C1_session_note.txt", "r", encoding="utf-8") as input_file:
    saved_message = input_file.read()

print(saved_message)
````

`input_file.read()` and `output_file.write(...)` are **method calls**: a method is a function attached to an object. You already used methods with lists (`append`, `pop`) and dictionaries (`get`, `items`).  
`input_file.read()`和`output_file.write(...)`是**方法调用**:一个方法是一个附着在物体上的函数. 您已经使用列表( `append`, `pop`) 和字典( `get`, `items`) 的方法。

> Text files contain strings. Numeric text must be converted before calculation. Later courses introduce NumPy tools for efficiently reading and writing numerical arrays.

概念图：**可变性和类型 容器 条件 环 格式文本 文件输出**.


````python
# 1. Variables and containers
operator = "Student A"  # Optional: replace with input("Operator name: ")
valid_range_N = (0.0, 100.0)  # tuple: fixed lower and upper limits

session = {
    "test_name": "spring loading",
    "operator": operator,
    "unit": "N",
    "measurements": [12.4, 18.7, -2.0, 24.1, 105.0, 31.6],
}

print(session)
````

### Your task｜你的任务

Before running the solution, identify which measurements are outside the valid interval. Then explain why `valid_measurements` is a list, `valid_range_N` is a tuple, and `session` is a dictionary.  
在运行溶液之前，确定哪些测量值位于有效间隔之外. 那就解释为什么`valid_measurements`是一个列表，`valid_range_N`是一个Tuple,`session`是一个字典。


````python
# 2. Loop over the data and use a condition
valid_measurements = []
rejected_measurements = []
lower_limit, upper_limit = valid_range_N

for number, force in enumerate(session["measurements"], start=1):
    if lower_limit <= force <= upper_limit:
        valid_measurements.append(force)
        status = "valid"
    else:
        rejected_measurements.append(force)
        status = "rejected"
    print(f"Measurement {number}: {force:.1f} N -> {status}")

# 3. Calculate and store derived results
if valid_measurements:
    mean_force = sum(valid_measurements) / len(valid_measurements)
else:
    mean_force = None

session["valid_measurements"] = valid_measurements
session["rejected_measurements"] = rejected_measurements
session["mean_force_N"] = mean_force
````

````python
# 4. Build and save a readable report
report_lines = [
    f"Test: {session['test_name']}",
    f"Operator: {session['operator']}",
    f"Valid measurements: {len(valid_measurements)}",
    f"Rejected measurements: {len(rejected_measurements)}",
    f"Mean valid force: {mean_force:.2f} {session['unit']}",
]
report = "\n".join(report_lines)

with open("C1_measurement_report.txt", "w", encoding="utf-8") as report_file:
    report_file.write(report)

print("\n" + report)
````

## Conclusion: need for additional libraries｜结论：为什么需要扩展库

Python's standard library provides general-purpose tools, but scientific work benefits from specialised libraries imported with `import`. In particular, basic lists are not designed for fast vectorised numerical calculations.  
Python的标准库提供通用工具，但科学工作受益于用`import`导入的专门库. 尤其是，基本列表不是为快速向量数字计算而设计的。

The main libraries for scientific computing are **NumPy** and **SciPy**, which provide mathematical functions and tools for arrays, linear algebra, differential equations, optimisation, and more.  
科学计算的主要库有**NumPy**和**SciPy**，它们为数组，线性代数，微分方程，优化等提供了数学函数和工具。

The standard plotting library is **Matplotlib**.
