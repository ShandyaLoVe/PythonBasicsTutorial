## 1 — Images as pixel arrays｜1 — 图像作为像素数组

In digital image processing, an image is modelled as an array of **pixels**. Each pixel contains one or more numerical values describing brightness or colour. Array indexing therefore identifies spatial positions in the image.  
在数字图像处理中，图像被表示为**像素**数组。每个像素包含一个或多个表示亮度或颜色的数值，因此数组索引也对应着图像中的空间位置。

### Grayscale images｜灰度图像

A common 8-bit grayscale image has shape $(M,N)$ and integer values from 0 (black) to 255 (white). Other image formats may use floating-point values, commonly from 0 to 1, so the `dtype` and range should always be inspected rather than assumed.  
常见的 8 位灰度图形状为 $(M,N)$，整数值从 0（黑）到 255（白）。其他格式也可能使用浮点数，通常位于 0 到 1 之间，因此不能只凭经验假设，必须检查 `dtype` 与数值范围。

The example below creates a small $6\times6$ grayscale image and displays it with `imshow`. Setting `vmin=0` and `vmax=255` fixes the brightness scale, which is important when comparing several processed versions.  
下面创建一幅 $6\times6$ 灰度图，并用 `imshow` 显示。显式设置 `vmin=0`、`vmax=255` 可以固定亮度尺度，这在比较多幅处理结果时非常重要。


```python
import numpy as np
import matplotlib.pyplot as plt

# A 6 × 6 grayscale image stored as 8-bit unsigned integers.
image = np.array([
    [  0,  51, 102, 153, 204, 255],
    [ 51,   0,  51, 102, 153, 204],
    [102,  51,   0,  51, 102, 153],
    [153, 102,  51,   0,  51, 102],
    [204, 153, 102,  51,   0,  51],
    [255, 204, 153, 102,  51,   0],
], dtype=np.uint8)

print("Shape:", image.shape, "dtype:", image.dtype,
      "range:", (image.min(), image.max()))

plt.figure(1, figsize=(5, 4))
plt.imshow(image, cmap="gray", vmin=0, vmax=255, origin="upper")
plt.colorbar(label="grayscale intensity")
plt.title("A 6 × 6 grayscale image")
plt.xlabel("column / x")
plt.ylabel("row / y")
plt.tight_layout()
```

With Matplotlib's default `origin='upper'`, image coordinates follow the usual matrix convention:

- columns form the horizontal display coordinate $x$ and increase from left to right;
- rows form the vertical display coordinate $y$ and increase from top to bottom;
- the top-left pixel has array index `[0, 0]`.

在 Matplotlib 默认的 `origin='upper'` 设置下，图像坐标遵循矩阵习惯：列对应横向显示坐标 $x$，从左向右增加；行对应纵向显示坐标 $y$，从上向下增加；左上角像素的数组索引为 `[0, 0]`。

All familiar array techniques remain available. A single index changes one pixel, a slice changes a row, column, or rectangle, and a Boolean mask changes pixels selected by their values. The next example uses `.copy()` so each operation starts from the same original image.  
此前学习的数组工具都可以直接使用：单个索引修改一个像素，切片修改一行、一列或一个矩形区域，布尔掩码则按像素值筛选和修改。下面使用 `.copy()`，保证每种操作都从同一幅原图开始。


```python
im1 = image.copy()
im1[0, 5] = 0          # one pixel

im2 = image.copy()
im2[3, :] = 255        # one complete row

im3 = image.copy()
im3[:, 4] = 255        # one complete column

im4 = image.copy()
im4[3:, 4:] = 100      # a rectangular block

im5 = image.copy()
bright_mask = im5 > 150
im5[bright_mask] = 255 # threshold selected pixels to white

images = [image, im1, im2, im3, im4, im5]
titles = ["Original", "One pixel [0, 5]", "One row [3, :]",
          "One column [:, 4]", "Rectangle [3:, 4:]", "Threshold > 150"]

plt.figure(2, figsize=(15, 3.5))
for position, (current_image, title) in enumerate(zip(images, titles), start=1):
    plt.subplot(1, 6, position)
    plt.imshow(current_image, cmap="gray", vmin=0, vmax=255, origin="upper")
    plt.title(title, fontsize=10)
    plt.xlabel("column")
    if position == 1:
        plt.ylabel("row")
plt.tight_layout()
```

### Colour images in RGB format｜RGB 彩色图像

An RGB pixel contains three channel values: red, green, and blue. An $M\times N$ colour image is therefore stored as one array of shape $(M,N,3)$. The last axis selects the channel:

- `image_rgb[:, :, 0]` — red channel;
- `image_rgb[:, :, 1]` — green channel;
- `image_rgb[:, :, 2]` — blue channel.

RGB 像素包含红、绿、蓝三个通道值，因此一幅 $M\times N$ 彩色图通常保存为形状 $(M,N,3)$ 的数组。最后一个轴用于选择颜色通道：索引 0、1、2 分别对应红、绿、蓝。

For an 8-bit RGB image, each channel normally uses integers from 0 to 255. The code below constructs a $2\times5$ image by stacking three two-dimensional channel arrays and then displays both the combined image and the individual channel intensities.  
对于 8 位 RGB 图像，每个通道通常使用 0 到 255 的整数。下面把三个二维通道数组堆叠成一幅 $2\times5$ 彩色图，并分别显示合成图与各通道强度。


```python
# Three 2 × 5 colour-channel arrays.
R = np.array([[255,   0, 128,  64, 192],
              [  0, 255,   0, 128,  64]], dtype=np.uint8)
G = np.array([[  0, 255,  64, 192, 128],
              [128,   0, 255,  64, 192]], dtype=np.uint8)
B = np.array([[ 64, 128, 255,   0, 192],
              [192,  64, 128, 255,   0]], dtype=np.uint8)

# Stack along the last axis: (2, 5) + three channels → (2, 5, 3).
image_rgb = np.stack((R, G, B), axis=-1)
print("RGB image shape:", image_rgb.shape, "dtype:", image_rgb.dtype)

plt.figure(3, figsize=(14, 4))
plt.subplot(1, 4, 1)
plt.imshow(image_rgb)
plt.title("RGB image")

for position, (channel, colour_map, title) in enumerate(
    [(R, "Reds", "Red intensity"),
     (G, "Greens", "Green intensity"),
     (B, "Blues", "Blue intensity")], start=2
):
    plt.subplot(1, 4, position)
    plt.imshow(channel, cmap=colour_map, vmin=0, vmax=255)
    plt.title(title)

plt.tight_layout()
```

## 2 — Loading and displaying an image file｜2 — 读取并显示图像文件

Real applications usually start from an existing image rather than a hand-written pixel array. `plt.imread(path)` reads a supported image file and returns a NumPy array. Before processing it, inspect:

- `shape`: image height, width, and number of channels;
- `dtype`: integer or floating-point storage;
- minimum and maximum values: the actual numerical range.

实际应用通常从已有图像开始，而不是手工填写像素。`plt.imread(path)` 会读取支持的图像文件并返回 NumPy 数组。开始处理前，应检查图像高度、宽度与通道数，确认整数或浮点存储类型，并查看真实的最小值与最大值。

The next cell loads `nautilus.png` or `nautile.png` if available. Since neither is included in the present project, it otherwise creates a spiral-like RGB test image. This fallback preserves the complete import workflow while keeping the notebook executable.  
下一个单元会优先读取 `nautilus.png` 或 `nautile.png`。当前项目未包含这两个文件，因此会改为生成一幅带螺旋纹理的 RGB 测试图；这样既保留完整的导入流程，又保证 notebook 可以运行。


```python
from pathlib import Path

# Accept both the English filename from the handout and the French filename
# used by the original code.
image_candidates = [Path("nautilus.png"), Path("nautile.png")]
image_path = next((candidate for candidate in image_candidates if candidate.exists()), None)

if image_path is not None:
    tab_image = plt.imread(image_path)
    image_source = f"loaded from {image_path.name}"
else:
    # Deterministic RGB fallback with a spiral-like pattern.
    rows, columns = 240, 320
    yy, xx = np.mgrid[0:rows, 0:columns]
    centre_x, centre_y = columns / 2, rows / 2
    radius = np.hypot(xx - centre_x, yy - centre_y)
    angle = np.arctan2(yy - centre_y, xx - centre_x)
    spiral = 0.5 + 0.5 * np.cos(0.20 * radius - 3 * angle)
    radial_fade = np.clip(1 - radius / 210, 0, 1)
    red = np.clip(0.25 + 0.75 * spiral * radial_fade, 0, 1)
    green = np.clip(0.15 + 0.65 * radial_fade, 0, 1)
    blue = np.clip(0.30 + 0.70 * (1 - spiral) * radial_fade, 0, 1)
    tab_image = np.stack((red, green, blue), axis=-1)
    image_source = "generated fallback (original image file not found)"

print("Source:", image_source)
print("Shape:", tab_image.shape)
print("dtype:", tab_image.dtype)
print("Value range:", (float(tab_image.min()), float(tab_image.max())))

plt.figure(4, figsize=(8, 5))
plt.imshow(tab_image, origin="upper")
plt.xlabel("column / x", fontsize=13)
plt.ylabel("row / y", fontsize=13)
plt.title("Loaded or generated RGB image")
plt.tight_layout()
```

### Image coordinates and array indices｜图像坐标与数组索引

If an imported RGB image has shape `(429, 462, 3)`, it contains 429 rows, 462 columns, and three colour channels. Valid row indices are 0–428 and valid column indices are 0–461. However, not every reader returns values from 0 to 255: PNG images read by Matplotlib commonly use floating-point values from 0 to 1.  
如果导入的 RGB 图像形状为 `(429, 462, 3)`，它包含 429 行、462 列和三个颜色通道；有效行索引为 0–428，有效列索引为 0–461。但像素范围并不总是 0–255，例如 Matplotlib 读取 PNG 时通常返回 0–1 的浮点数。

The top-left point is the **origin**, not the centre. Matplotlib reports display coordinates in the order $(x,y)=(\text{column},\text{row})$, whereas NumPy indexes an image as `image[row, column]`. Therefore a cursor position near $(x,y)=(271,279)$ corresponds to `image[279, 271]`. This distinction matters whenever distances or pixel values are measured.  
左上角是坐标**原点**，不是中心。Matplotlib 按 $(x,y)=(\text{列},\text{行})$ 显示坐标，而 NumPy 使用 `image[行, 列]` 索引。因此光标位置 $(x,y)=(271,279)$ 对应 `image[279, 271]`。测量距离或读取像素值时必须注意这一顺序差异。

The original screenshot reported an almost-white pixel with RGB values close to `(255, 254, 255)`. In a floating-point 0–1 image, the equivalent is approximately `(1.0, 0.996, 1.0)`. The screenshots referenced by the original handout are not included in this project, so the revised notebook demonstrates the same coordinate ideas directly on the generated or loaded array.  
原截图记录了一个接近白色的像素，其 RGB 约为 `(255, 254, 255)`；若图像采用 0–1 浮点范围，对应值约为 `(1.0, 0.996, 1.0)`。原讲义引用的截图同样未包含在当前项目中，因此修订版直接在已读取或生成的数组上演示相同的坐标概念。


### Locating a pixel interactively｜交互式定位像素

In an interactive Matplotlib window, moving the pointer over an image displays approximate $x$, $y$, and colour values in the status bar. Round $x$ to a column index and $y$ to a row index, then access the pixel as `image[row, column]`. Always check that the rounded indices remain inside the image bounds.  
在 Matplotlib 交互窗口中，移动鼠标时，状态栏会显示近似的 $x$、$y$ 坐标和颜色值。把 $x$ 四舍五入为列索引，把 $y$ 四舍五入为行索引，再使用 `image[row, column]` 读取像素；同时要确认索引没有超出图像范围。

For reproducible analysis, programmatic indices are preferable to manually copied cursor values. The integrated exercise below selects a central rectangle, calculates brightness, and uses a mask to highlight bright pixels.  
为了让分析可复现，程序化索引通常比手工抄录鼠标坐标更可靠。下面的综合练习会提取中心矩形、计算亮度，并使用掩码突出较亮像素。

## 3 — Integrated practice: crop and threshold an image｜3 — 综合练习：裁剪图像并进行亮度筛选

This example connects the whole sequence of courses: NumPy represents the data, Course 4 slicing extracts the central rectangle, Course 5 creates a brightness mask, and Matplotlib displays the result. The imported image is first converted to floating-point RGB values from 0 to 1 so that the threshold has a consistent meaning.  
这个示例把前几课串联起来：NumPy 表示数据，第 4 课的切片提取中心矩形，第 5 课的布尔掩码筛选亮像素，Matplotlib 则负责显示结果。首先把导入图像统一转换为 0 到 1 的浮点 RGB 数据，使亮度阈值具有一致含义。


```python
rgb = np.asarray(tab_image[..., :3], dtype=float)
if rgb.max() > 1:
    rgb = rgb / 255.0

height, width = rgb.shape[:2]
row_start, row_stop = height // 4, 3 * height // 4
col_start, col_stop = width // 4, 3 * width // 4
central_crop = rgb[row_start:row_stop, col_start:col_stop]

# A simple brightness estimate: mean of the R, G, and B channels.
brightness = central_crop.mean(axis=2)
bright_mask_crop = brightness >= 0.65
highlighted_crop = central_crop.copy()
highlighted_crop[~bright_mask_crop] *= 0.2

print("Full RGB shape:", rgb.shape)
print("Crop shape:", central_crop.shape)
print(f"Bright-pixel fraction: {bright_mask_crop.mean():.1%}")

plt.figure(5, figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(rgb)
plt.title("Complete image")
plt.xlabel("column / x")
plt.ylabel("row / y")

plt.subplot(1, 3, 2)
plt.imshow(central_crop)
plt.title("Central crop")
plt.xlabel("column inside crop")

plt.subplot(1, 3, 3)
plt.imshow(highlighted_crop)
plt.title("Pixels with brightness ≥ 0.65")
plt.xlabel("column inside crop")

plt.tight_layout()
```
