## 1 — Defining and calling a function｜1 — 定义并调用函数

A function groups several instructions under one name. A `def` statement creates the function object but does not execute its body. The indented body runs only when the function is **called** with parentheses. Functions are often placed near the beginning of a script or in a separate module so they can be reused.  
函数把若干条指令组织在同一个名称下。`def` 语句会创建函数对象，但不会立即执行函数体；只有使用圆括号**调用**函数时，缩进的函数体才会运行。函数通常放在脚本前部，或放入独立模块以便复用。

The first example has no parameters and only prints the integers from 1 to 5.  
第一个示例没有参数，只负责输出 1 到 5。


```python
def count_to_five():
    """Print the integers from 1 through 5."""
    for number in range(1, 6):
        print(number)
```

The definition contains four important elements:

- `def`, followed by a valid function name;
- parentheses containing the parameters, empty in this example;
- a colon and an indented function body;
- a docstring between triple quotes, which `help()` can display.

函数定义包含四个重要部分：`def` 与合法函数名、用于填写形参的圆括号、冒号与缩进函数体，以及可由 `help()` 显示的三引号文档字符串。

Printing the name without parentheses shows that the variable refers to a function object. Calling `count_to_five()` executes the body. This difference between referring to a function and calling it will become important when functions themselves are passed as arguments.  
不加圆括号打印函数名，只会显示该变量指向一个函数对象；调用 `count_to_five()` 才会执行函数体。以后把函数本身作为参数传递时，这一区别尤其重要。


```python
print("Function object:")
print(count_to_five)

print()
print("Function help:")
help(count_to_five)

print("Function call:")
count_to_five()
```

### Parameters, arguments, and return values｜形参、实参与返回值

A **parameter** is a name in the function definition; an **argument** is the actual value supplied during a call. Most functions receive one or more arguments and send a result back with `return`. A returned value can be assigned to a variable, used in another expression, or passed directly to another function.  
**形参**是函数定义中使用的名称，**实参**是调用时传入的实际值。多数函数接收一个或多个实参，并通过 `return` 把结果送回调用位置；返回值可以保存到变量中，也可以继续参与表达式或传给另一个函数。

`return` ends the current function call. A function with no explicit `return` returns `None`. Parameters, accepted input types, return values, units, and important assumptions should be stated in the docstring.  
`return` 会结束当前函数调用。没有显式 `return` 的函数会返回 `None`。文档字符串应说明参数、可接受的数据类型、返回值、单位以及重要假设。


```python
import numpy as np

def square(x):
    """Return x squared element by element.

    Parameters
    ----------
    x : number or NumPy array

    Returns
    -------
    number or NumPy array
        ``x ** 2`` with the same array shape as x.
    """
    result = x**2
    return result

print("Function help:")
help(square)

scalar_squared = square(4.0)
vector = np.array([1, 2, 3])
vector_squared = square(vector)
matrix = np.array([[1.0, 2.0], [-2.0, 5.0]])
matrix_squared = square(matrix)

print("square(4.0) =", scalar_squared)
print("Vector and element-wise square:")
print(vector)
print(vector_squared)
print("Matrix and element-wise square:")
print(matrix)
print(matrix_squared)
```

### Multiple parameters and multiple return values｜多个参数与多个返回值

Parameters are separated by commas in the definition and arguments are separated by commas in the call. A function may return several values by separating them with commas; Python packages those values into a tuple.  
函数定义中的多个形参以及调用时的多个实参都使用逗号分隔。函数也可以用逗号分隔多个返回值，Python 会把它们组合成一个元组。

The caller can keep the tuple in one variable or unpack it into the same number of variables. The unpacking order must match the order used by `return`.  
调用者既可以用一个变量保存整个元组，也可以把它解包到数量相同的多个变量中；解包顺序必须与 `return` 中的顺序一致。


```python
def sum_and_product(x, y):
    """Return the element-wise sum and product of compatible inputs."""
    total = x + y
    product = x * y
    return total, product

vector1 = np.array([1, 2, 3])
vector2 = np.array([3, 4, 5])

both_results = sum_and_product(vector1, vector2)
print("One tuple containing both results:", both_results)

vector_sum, vector_product = sum_and_product(vector1, vector2)
print("Unpacked sum:", vector_sum)
print("Unpacked product:", vector_product)
```

## 2 — Local and global variables｜2 — 局部变量与全局变量

A name created inside a function is normally **local** to that call and cannot be accessed after the function returns. Parameters are local names too. A name defined at notebook or module level is **global** and can be read inside a function if no local name hides it.  
函数内部创建的名称通常只在本次调用中有效，属于**局部变量**，函数结束后不能从外部直接访问；形参同样是局部名称。在 notebook 或模块顶层定义的名称属于**全局变量**，如果没有被同名局部变量遮蔽，函数内部可以读取它。

Although reading a global constant can be convenient, hidden dependencies make functions harder to test and reuse. Passing required information explicitly as an argument is usually clearer. The examples below demonstrate both approaches.  
读取全局常量有时很方便，但隐藏依赖会降低函数的可测试性和复用性。通常把所需信息显式作为实参传入会更清晰；下面会同时演示两种写法。


```python
a = 2.0  # global value

def multiply_by_global_a(value):
    """Multiply by the global value a (works, but has a hidden dependency)."""
    result = a * value
    return result

def multiply(value, factor):
    """Multiply value by an explicit factor."""
    result = factor * value
    return result

print("Using global a:", multiply_by_global_a(5.0))
print("Using an explicit argument:", multiply(5.0, factor=2.0))
```

```python
try:
    print(result)  # result is local to the previous function calls
except NameError as error:
    print("Expected NameError for local variable 'result':", error)
```

```python
try:
    print(value)  # value is a local parameter name
except NameError as error:
    print("Expected NameError for local parameter 'value':", error)
```

The first function can read the global value `a`, so it returns the expected result. However, `y` and the parameter name `x` exist only inside their function call. Attempting to access them outside raises `NameError`; the revised cells catch and display that expected exception so the notebook can continue running.  
第一个函数能够读取全局变量 `a`，因此返回预期结果；但 `y` 和形参名称 `x` 只存在于函数调用内部，从外部访问会触发 `NameError`。修订后的代码会捕获并显示这个预期异常，使 notebook 能继续运行。

Compare `multiply_by_global_a(value)` with `multiply(value, factor)`. The second version exposes every dependency through its interface, so changing the factor does not depend on external state.  
请比较 `multiply_by_global_a(value)` 与 `multiply(value, factor)`：第二种写法通过接口显式呈现全部依赖，因此改变倍数时不依赖外部状态。

## 3 — Optional parameters｜3 — 可选参数

A parameter can receive a default value in the definition, for example `title=None`. During a call, the default is used when no corresponding argument is provided; a keyword argument such as `title="My graph"` overrides it. Required parameters must appear before parameters with defaults.  
形参可以在定义中获得默认值，例如 `title=None`。调用时若没有提供对应实参，就使用默认值；使用 `title="My graph"` 这样的关键字实参则可覆盖默认值。必需形参必须写在带默认值的形参之前。

Prefer immutable defaults such as numbers, strings, tuples, or `None`. A mutable default such as `items=[]` is created only once and may unexpectedly retain changes between calls. When a new list is needed for every call, use `items=None` and create the list inside the function.  
默认值最好使用数字、字符串、元组或 `None` 等不可变对象。`items=[]` 这样的可变默认值只会创建一次，可能在多次调用之间意外保留修改；若每次调用都需要新列表，应使用 `items=None`，再在函数内部创建列表。

The following plotting function uses optional labels and a title to avoid repeating the same formatting commands. It returns the plotted line so callers can further customise it if needed.  
下面的绘图函数通过可选的坐标标签和标题避免重复格式设置，并返回绘制得到的线对象，方便调用者继续定制。


```python
import matplotlib.pyplot as plt

def plot_clean_graph(x, y, xlabel=r"$x$", ylabel=r"$y$", title=None,
                     color="black"):
    """Plot y against x with consistent labels and grid formatting."""
    line, = plt.plot(x, y, color=color)
    plt.grid(alpha=0.3)
    if title is not None:
        plt.title(title, fontsize=14)
    plt.xlabel(xlabel, fontsize=14)
    plt.ylabel(ylabel, fontsize=14)
    return line

x_values = np.linspace(0, 1, 101)
y_parabola = 1 - x_values**2
y_exponential = np.exp(x_values)

plt.figure(1, figsize=(12, 4))
plt.subplot(1, 2, 1)
plot_clean_graph(x_values, y_parabola)

plt.subplot(1, 2, 2)
plot_clean_graph(x_values, y_exponential, ylabel=r"$e^x$",
                 title="Exponential function", color="tab:blue")
plt.tight_layout()
```

## 4 — Passing functions as arguments｜4 — 将函数作为参数传递

Functions are ordinary Python objects. They can call previously defined functions, be stored in variables, and be passed as arguments. Pass `np.sin`, not `np.sin()`: the first is the function object, whereas the second attempts to call it immediately.  
函数也是普通 Python 对象，可以调用先前定义的函数、保存在变量中，也可以作为实参传递。应传入 `np.sin` 而不是 `np.sin()`：前者是函数对象，后者会尝试立即调用函数。

The next example receives a callable `function`, samples it over a chosen interval, and delegates the formatting to `plot_clean_graph`. The supplied function must accept a NumPy array and return compatible values; this property is often called vectorisation.  
下面的示例接收可调用对象 `function`，在指定区间进行采样，再把格式设置交给 `plot_clean_graph`。传入函数必须能够接收 NumPy 数组并返回形状兼容的结果，这正是前面学习的向量化能力。


```python
def plot_function(function, xmin=0, xmax=1, name="f"):
    """Sample and plot a vectorised function on [xmin, xmax]."""
    x_values = np.linspace(xmin, xmax, 200)
    y_values = function(x_values)
    title = f"Function {name} on [{xmin:g}, {xmax:g}]"
    return plot_clean_graph(x_values, y_values, title=title)

def bell_shape(x):
    """Return the bell-shaped polynomial x(1-x)."""
    return x * (1 - x)

plt.figure(2, figsize=(15, 4))
plt.subplot(1, 3, 1)
plot_function(bell_shape, name=r"$x(1-x)$")

plt.subplot(1, 3, 2)
plot_function(np.sin, name="sin")

plt.subplot(1, 3, 3)
plot_function(np.sin, xmax=4 * np.pi, name="sin")
plt.tight_layout()
```

## 5 — Integrated practice: package a mechanics workflow｜5 — 综合练习：封装力学计算流程

The final example defines a damped-oscillator model and passes that function to an analysis function. The analysis function evaluates the model, builds a Boolean mask for large displacements, and returns both the complete signal and the mask. NumPy performs the vectorised calculation, the mask selects important samples, and Matplotlib presents the result.  
最后的示例定义阻尼振动模型，并把这个函数传给分析函数。分析函数负责计算完整信号、构造大位移布尔掩码，再同时返回信号和掩码。NumPy 完成向量化计算，掩码筛选重要采样点，Matplotlib 展示结果，从而把本套课程的主要工具组织成可复用流程。


```python
def damped_position(time_s, amplitude_mm=20.0, damping_per_s=0.45,
                    frequency_hz=1.2):
    """Return the displacement of a damped oscillator in millimetres."""
    envelope = amplitude_mm * np.exp(-damping_per_s * time_s)
    return envelope * np.cos(2 * np.pi * frequency_hz * time_s)

def select_large_response(model, time_s, threshold_mm=5.0):
    """Evaluate a model and return values whose magnitude reaches a threshold."""
    values_mm = model(time_s)
    selected_mask = np.abs(values_mm) >= threshold_mm
    return values_mm, selected_mask

time_s = np.linspace(0, 5, 500)
position_mm, large_response_mask = select_large_response(
    damped_position, time_s, threshold_mm=5.0
)

print("Returned array shape:", position_mm.shape)
print("Selected sample count:", np.count_nonzero(large_response_mask))

plt.figure(3, figsize=(9, 4.5))
plot_clean_graph(time_s, position_mm, xlabel="Time (s)",
                 ylabel="Displacement (mm)", title="Damped oscillator",
                 color="navy")
plt.plot(time_s[large_response_mask], position_mm[large_response_mask],
         linestyle="none", marker=".", color="darkorange",
         label=r"$|x| \geq 5$ mm")
plt.legend()
plt.tight_layout()
```

### Function-design checklist｜函数设计检查表

1. Does the name describe one clear responsibility?｜函数名是否表达单一而明确的职责？
2. Are all required dependencies explicit parameters?｜所有必要依赖是否都通过参数明确传入？
3. Does the docstring describe inputs, outputs, units, and assumptions?｜文档是否说明输入、输出、单位和假设？
4. Are defaults immutable and meaningful?｜默认值是否安全且含义明确？
5. Does the function return data rather than only print it?｜函数是否返回数据，而不仅仅负责打印？
6. Can it be tested with a small predictable example?｜能否用一个结果可预测的小例子测试？
