## 1 — Creating matrices and inspecting array properties｜1 — 创建矩阵与检查数组属性

### Creation functions: `zeros`, `ones`, `eye`, and `diag`｜创建函数：`zeros`、`ones`、`eye` 与 `diag`

`np.zeros(shape)` and `np.ones(shape)` create arrays filled with zeros or ones. A matrix shape is supplied as a tuple such as `(3, 3)`. `np.eye(n)` creates the $n\times n$ identity matrix, while `np.diag(values)` places the supplied values on the main diagonal.

`np.zeros(shape)` 与 `np.ones(shape)` 分别创建全零和全一数组；矩阵形状使用 `(3, 3)` 这样的元组表示。`np.eye(n)` 创建 $n\times n$ 单位矩阵，`np.diag(values)` 则把给定数值放到主对角线上。


```python
import numpy as np

print("3x3 matrix of 1 \n",np.ones((3,3)))            # We use the tuple (3,3) to give the form of
                                                        # the matrix            

print("3x3 identity matrix \n",np.eye(3))             # For np.eye which creates the NxN identity matrix, a single argument 

print("Diagonal matrix \n",np.diag([3.,4.,-1.,6.]))  # the diagonal vector is the argument 
                                                        # the dimension of the matrix (square) adapts
```

To support the following examples, we create two $3\times3$ matrices $A$ and $B$. Matrix $A$ is a magic square containing the integers 1 to 9: every row, column, and main diagonal has the same sum. Matrix $B$ contains reproducible pseudo-random integers from 1 through 9.

为了支持后续示例，我们创建两个 $3\times3$ 矩阵 $A$ 与 $B$。$A$ 是由 1 到 9 组成的幻方，每行、每列以及两条主对角线的和相同；$B$ 则由 1 到 9 的可复现伪随机整数组成。固定随机种子后，全班都能获得相同结果，便于比较与调试。


```python
A = np.array([[8, 1, 6],
              [3, 5, 7],
              [4, 9, 2]])

rng = np.random.default_rng(42)  # fixed seed for reproducible lessons
B = rng.integers(1, 10, size=(3, 3))

print("A =\n", A)
print("B =\n", B)
```

### Characteristics of a table: dimensions, axes, size, number of elements, shape｜数组属性：维数、轴、大小、元素数量与形状

The function `np.ndim` allows you to know the number of dimensions of the table. 1 for a vector, 2 for a matrix.  
函数 `np.ndim` 允许您知道数组的维数。 1表示向量，2表示矩阵。


```python
print("Number of dimensions of A: ", np.ndim(A)) 
```

For a matrix, **axis 0** runs down the rows and **axis 1** runs across the columns. When a reduction such as `sum` is performed *along* an axis, that axis is collapsed: summing along axis 0 produces one result per column, whereas summing along axis 1 produces one result per row.

对于矩阵，**轴 0** 沿行号向下，**轴 1** 沿列号向右。当 `sum` 等聚合运算沿某个轴执行时，该轴会被消去：沿轴 0 求和得到每一列的结果，沿轴 1 求和得到每一行的结果。

`np.size(A)` gives the total number of elements. Supplying `axis=0` or `axis=1` gives the length of the corresponding axis.

`np.size(A)` 返回元素总数；指定 `axis=0` 或 `axis=1` 则返回相应轴的长度。


```python
print("Total number of items: ", np.size(A)) 
print("Number of elements along axis 0: ", np.size(A,axis=0)) 
print("Number of elements along axis 1: ", np.size(A,1))  # We can omit the 'axis =' 
```

`np.shape(A)` returns a tuple containing the length of every axis. For a matrix, `(number_of_rows, number_of_columns)` is the most important information to check before an operation. The equivalent attribute is `A.shape`.

`np.shape(A)` 返回由各轴长度组成的元组。对于矩阵，结果依次表示“行数、列数”，因此在运算前检查形状非常重要；也可以使用等价属性 `A.shape`。


```python
print("The dimensions of table A are: ", np.shape(A)) 
```

## 2 - Operations on matrices: component by component, matrix product, calculation on the components｜2 - 矩阵运算：逐元素运算、矩阵乘法与元素统计

### Operations component by component｜逐元素运算

As we have already seen, we can use the classic operators of addition, subtraction, division or multiplication between $A$ and $B$ to perform component by component operations:  
正如我们已经看到的那样，我们可以使用$A$和$B$之间加法、减法、除法或乘法的经典运算符来逐元素操作执行组件：


```python
print("A + B = \n", A+B)
```

The addition succeeds because $A$ and $B$ have the same shape. Internally, compiled NumPy code visits the elements efficiently; we express the intended array operation instead of writing nested Python loops ourselves.

由于 $A$ 与 $B$ 形状相同，加法可以逐元素进行。NumPy 会在底层高效遍历元素，我们只需要表达数组层面的运算，而不必亲自编写多层 Python 循环。

> Vectorisation does not mean that loops disappear—it means that the low-level loops are handled by NumPy. A Python loop is still appropriate when an algorithm cannot naturally be expressed as array operations.  
> 向量化并不意味着循环从此消失，而是把底层循环交给 NumPy。若某个算法无法自然地写成数组运算，Python 循环依然是合理选择。


```python
print("A * B = \n", A*B)

print("A / B = \n", A / B)
```

As expected, `A * B` and `A / B` return element-wise products and quotients. This is different from the matrix product used in linear algebra.

正如预期，`A * B` 与 `A / B` 分别返回逐元素的乘积与商；这与线性代数中的矩阵乘法不同。

### Matrix–matrix and matrix–vector products｜矩阵–矩阵与矩阵–向量乘法

Use `A @ B` for matrix multiplication; `np.dot(A, B)` gives the same result for two-dimensional arrays. Be careful: `A * B` is element-wise, whereas `A @ B` is a matrix product.

矩阵乘法使用 `A @ B`；对于二维数组，`np.dot(A, B)` 给出相同结果。务必区分：`A * B` 是逐元素乘法，`A @ B` 才是矩阵乘法。

If $A$ has shape $(m_A,n_A)$ and $B$ has shape $(m_B,n_B)$, the product exists only when $n_A=m_B$ and its shape is $(m_A,n_B)$:

$$ (AB)_{ij}=\sum_{k=1}^{n_A}A_{ik}B_{kj}. $$

也就是说，左矩阵的列数必须等于右矩阵的行数；求和指标 $k$ 同时遍历 $A$ 的列与 $B$ 的行，外侧保留的两个维度决定结果形状。


```python
print("Matrix product using np.dot:\n", np.dot(A, B))
print("The same product using @:\n", A @ B)

Ared = A[:2, :]  # first two rows: shape (2, 3)
Bred = B[:, :2]  # first two columns: shape (3, 2)

print("Ared shape:", Ared.shape, "Bred shape:", Bred.shape)
print("Product of rectangular matrices, shape (2, 2):\n", Ared @ Bred)
```

The matrix–vector product $Mv$ is a special case. If $M$ has shape $(m,n)$ and the one-dimensional array `v` has shape `(n,)`, then `M @ v` returns a one-dimensional result of shape `(m,)`.

矩阵–向量乘法 $Mv$ 是矩阵乘法的特例。若 $M$ 的形状为 $(m,n)$，一维数组 `v` 的形状为 `(n,)`，那么 `M @ v` 返回形状为 `(m,)` 的一维结果。

NumPy's one-dimensional arrays do not separately encode “row vector” and “column vector”. If a true column-shaped object is needed, use a two-dimensional array of shape `(n, 1)`. This explains why `A @ v` and `A @ vcol` contain related values but have different shapes.

NumPy 的一维数组不额外区分“行向量”和“列向量”。若确实需要列形状，应使用 `(n, 1)` 的二维数组。因此 `A @ v` 与 `A @ vcol` 的数值相关，但结果形状不同。


```python
v = np.array([1, 2, 3]) # v is an array of dimension 1 = a vector, of size 3

print("Matrix-vector product:\n", A @ v)

vcol = np.array([[1],
                 [2],
                 [3]]) # vcol is an array of dimension 2 = a matrix, 
                       # of dimensions 3x1: it is a "column vector"

print("Column matrix-vector product:\n", A @ vcol)
```

### Reduction functions: `sum`, `mean`, `max`, `min`, `argmax`, and `argmin`｜聚合函数

A reduction combines many array elements into fewer results. `np.sum(A)` sums every element; supplying an `axis` performs one reduction for each column or row. Always predict which axis disappears and what the remaining shape will be.

聚合运算把多个数组元素合并为较少的结果。`np.sum(A)` 对所有元素求和；指定 `axis` 后，则分别对各列或各行聚合。使用前应先判断哪一个轴会消失，以及结果的形状。


```python
print("As a reminder, matrix A is: \n",A)
```

```python
np.sum(A) # sum of all elements of A.  
```

```python
np.sum(A,0) # returns a vector which is the sum along axis 0 
            # i.e. the sum of each column 8+3+4.1+5+9.6+7+2 
```

```python
np.sum(A,1) # returns a vector which is the sum along axis 1 i.e. the sum of each line 
```

```python
np.diag(A) # returns a vector containing the elements of the diagonal of A
```

```python
np.sum(np.diag(A)) # chaining the two functions we obtain the sum of the elements of the diagonal of A
```

`np.mean`, `np.min`, and `np.max` follow the same axis rules as `np.sum`. Without an `axis`, `np.argmin` and `np.argmax` return the position in flattened row-major order. Convert that flat index back to matrix coordinates with `np.unravel_index(index, A.shape)`.

`np.mean`、`np.min` 与 `np.max` 遵循和 `np.sum` 相同的轴规则。未指定 `axis` 时，`np.argmin` 与 `np.argmax` 返回数组按行展开后的一维位置；可以使用 `np.unravel_index(index, A.shape)` 将其转换回矩阵坐标。


```python
minimum_index = np.argmin(A)
maximum_index = np.argmax(A)

print("Mean of all elements:", np.mean(A))
print("Minimum:", np.min(A))
print("Flat index of minimum:", minimum_index)
print("Matrix coordinates of minimum:", np.unravel_index(minimum_index, A.shape))
print("Maximum:", np.max(A))
print("Flat index of maximum:", maximum_index)
print("Matrix coordinates of maximum:", np.unravel_index(maximum_index, A.shape))
```

## 3 - Extractions of elements for a matrix｜3 - 矩阵元素与子块的提取

To illustrate the following manipulations, we use the Vandermonde matrix $C$ seen in TD, here of size $4\times5$. There is a ready-made function for this in `numpy`. The detail of the matrix does not matter for this course (more information on this matrix in your mathematics course or on [wikipedia](https://fr.wikipedia.org/wiki/Matrice_de_Vandermonde))  
为了说明以下操作，我们使用了在TD中看到的范德蒙德矩阵$C$，这里是大小为$4\times5$。 `numpy`中有现成的函数。 矩阵的细节对本课程无关紧要(关于数学课程中或[wikipedia](https://fr.wikipedia.org/wiki/Matrice_de_Vandermonde)中此矩阵的更多信息)


```python
V = np.vander([1, 2, 3, 5], 5, increasing=True)
print("Vandermonde matrix V =\n", V)
print("Shape of V:", V.shape)
```

We can **extract a submatrix from `V` (which we will also call "block")** using the colon `:` for each of the two axes as we have already seen during the introduction to Numpy.  
我们可以** 从 `V`(我们也将称之为"块")** 中提取一个子矩阵，使用`:` 的结肠，用于两个轴的每个轴，就像我们在介绍NumPy时已经看到的那样。


```python
print("Submatrix of V: \n",V[0:4,2:5])  # Extract indices from 0 to 3 for the first axis
                                        # and 2 to 4 for the second. 
                                        # corresponds to rows 1 to 4 and columns 3 to 5 of the matrix
```

It is also possible to**set the pace** with the syntax `start:end:step` example `2:11:2` will take the elements 2,4,6,8,10.  
也可以**以`start:end:step`语法例`2:11:2`来设定速度**，取2,4,6,8,10元素。


```python
print('Even rows and columns:\n',V[0:5:2,0:5:2]) # We impose a step of 2
```

It is also useful to use -1, -2, -3, ... as indices to count from the end index of the array. You can also put nothing to designate the beginning or the end.  
也可以使用 -1, -2, -3,。..作为指数从数组的末端索引中计数。 您也可以不指定开始或结束。


```python
print("V[1:-1,:] = ")
print(V[1:-1,:]) # Extract second to penultimate element for first index (lines) "1:-1" 
                 # and all elements for the second index (the columns) with ":"  
```

```python
print("V[:-2,1:] = ")
print(V[:-2,1:])  # We start from 0 and stop at -2 from the end for the first index 
                  # and we start from 1 until the end for the second index. 
```

The [SciPy Lecture Notes section on indexing and slicing](http://scipy-lectures.org/intro/numpy/array_object.html#indexing-and-slicing) provides a useful visual summary of one- and two-dimensional slices. The image referenced by the original worksheet is not included in this project, so use the examples above as the executable version of the same ideas: `start:stop:step` applies independently along each axis, and every stop index is excluded.  
[SciPy Lecture Notes 的索引与切片章节](http://scipy-lectures.org/intro/numpy/array_object.html#indexing-and-slicing)提供了直观总结。原课件引用的图片未包含在当前项目中，因此以上可运行示例就是同一知识点的完整替代：`start:stop:step` 可分别作用于每个轴，并且停止索引始终不包含在结果中。


### Views, aliases, and independent copies｜视图、别名与独立副本

This point needs special care. `V2 = V1` creates another name for the same array. Basic slicing such as `V1[2:, 2:]` normally creates a **view**, so changing the view's elements also changes `V1`. To obtain independent data, call `.copy()` explicitly. Rebinding a variable, as in `block = 3`, only changes what the name refers to and does not modify the old array.

这里必须特别小心：`V2 = V1` 只是让两个变量指向同一个数组；`V1[2:, 2:]` 这样的基本切片通常产生**视图**，修改视图中的元素也会改变 `V1`。如果需要互不影响的数据，必须显式调用 `.copy()`。而 `block = 3` 只是让变量改为指向整数 3，并不会修改原数组。

> This corrects an inaccurate statement in the original handout, which described a basic slice as an automatic copy.  
> 这一说明修正了原始讲义中的不准确表述：NumPy 的基本切片并不会自动复制数据。


```python
# 1. Alias: two names refer to exactly the same array.
V1 = V.copy()
V2 = V1
V2[2:, 2:] = 3
print("After changing V2, V1 also changes (alias):\n", V1)

# 2. View: a basic slice shares data with its parent array.
V1 = V.copy()
V1_view = V1[2:, 2:]
V1_view[:] = 3
print("After changing a slice view, V1 also changes:\n", V1)

# 3. Copy: .copy() creates independent data.
V1 = V.copy()
V1_copy = V1[2:, 2:].copy()
V1_copy[:] = 3
print("After changing an independent copy, V1 is unchanged:\n", V1)
```

## 4 — Changing shape with `np.reshape`｜4 — 使用 `np.reshape` 改变形状

`np.reshape` rearranges the same elements into a new shape. **The number of elements must remain unchanged.** Since `V` contains 20 values, valid two-dimensional shapes include `(1, 20)`, `(2, 10)`, `(4, 5)`, `(5, 4)`, `(10, 2)`, and `(20, 1)`; `(20,)` is the corresponding one-dimensional shape.

`np.reshape` 可以把同一组元素重新排列为新的形状，但**元素总数必须保持不变**。`V` 含有 20 个元素，因此可以重塑为 `(1, 20)`、`(2, 10)`、`(4, 5)`、`(5, 4)`、`(10, 2)`、`(20, 1)` 等二维形状，也可以变为一维的 `(20,)`。

By default, NumPy reads and writes the elements in row-major order. Predict the new row boundaries before running the example.

默认情况下，NumPy 按行读取并重新排列元素。运行示例前，可以先预测新数组每一行的分界位置。


```python
print("V= \n",V)

print("V in vector of 20 elements:\n", np.reshape(V,(20)))

print("V in matrix with 10 rows and 2 columns:\n", np.reshape(V,(10,2)))
```

## 5 — Transposing, flipping, and rotating arrays｜5 — 转置、翻转与旋转数组

NumPy includes mathematical transformations such as `transpose`, as well as spatial operations such as `rot90`, `flipud`, and `fliplr`. Here `ud` means **up–down** and `lr` means **left–right**. Use `help(np.rot90)` or the function's documentation when you need its exact arguments.

NumPy 既提供 `transpose` 这样的数学变换，也提供 `rot90`、`flipud` 和 `fliplr` 等空间操作；其中 `ud` 表示上下，`lr` 表示左右。需要确认参数时，可以运行 `help(np.rot90)` 或查阅函数文档。

Transposition is common in matrix calculations. Flips and rotations become especially intuitive when an array represents an image, which prepares us for Course 6. Remember that `np.rot90(V)` rotates counter-clockwise by default.

转置常用于矩阵计算；当数组表示图像时，翻转和旋转会更加直观，这也为第 6 课的图像处理做准备。默认情况下，`np.rot90(V)` 按逆时针方向旋转 90°。


```python
print("As a reminder, matrix V:")
print(V)
```

```python
print("Transpose of V: ")
print(np.transpose(V)) 
```

```python
print("Reversal \"upside down\" (line reversal): ")
print(np.flipud(V)) # or np.flip(V,axis=0) or shortcut np.flip(V,0)  
                    # 0 designates the axis to choose for the flip
```

```python
print("Reversal \"left right\" (inversion of columns): ")
print(np.fliplr(V)) # or np.flip(V,1) or np.flip(V,axis=1)
```

```python
print("Matrix rotation: ")
print(np.rot90(V)) # rotates the matrix 90 degrees in the trigo direction.
```

## 6 — Integrated practice: a small sensor grid｜6 — 综合练习：小型传感器网格

A $4\times5$ array represents four sensors measured at five successive times. In one workflow we reshape raw data, inspect its shape, calculate row and column statistics, extract a block, locate the largest value, and transpose the grid. The same operations will later apply to images, where rows and columns represent pixel coordinates.

一个 $4\times5$ 数组表示 4 个传感器在 5 个连续时刻的测量值。下面将在同一流程中重塑原始数据、检查形状、计算行列统计量、提取子块、定位最大值并转置网格。以后处理图像时，行列将对应像素坐标，但数组操作完全相通。


```python
raw_measurements = np.arange(1, 21)
sensor_grid = raw_measurements.reshape(4, 5)

mean_per_sensor = sensor_grid.mean(axis=1)
mean_per_time = sensor_grid.mean(axis=0)
middle_block = sensor_grid[1:3, 1:4]
peak_flat_index = sensor_grid.argmax()
peak_position = np.unravel_index(peak_flat_index, sensor_grid.shape)
time_by_sensor = sensor_grid.T

print("Sensor grid, shape", sensor_grid.shape, ":\n", sensor_grid)
print("Mean per sensor (axis 1):", mean_per_sensor)
print("Mean per time (axis 0):", mean_per_time)
print("Middle block:\n", middle_block)
print("Peak position (row, column):", peak_position)
print("Transposed shape:", time_by_sensor.shape)
```
